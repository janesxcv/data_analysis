# Lab 00: Introduction to NumPy — Sales Data Analysis

В этой лабораторной я создаю массив ежедневных продаж четырёх товаров за 364 дня, считаю среднемесячные и недельные продажи и сравниваю скорость суммирования в Python и NumPy.

```python
import timeit
import numpy as np

np.random.seed(42)

num_days = 364
num_products = 4

daily_sales = np.random.randint(
    low=10,
    high=101,
    size=(num_days, num_products)
)

print("Размер массива:", daily_sales.shape)
print("Продажи за первые 5 дней:")
print(daily_sales[:5])
```

## 1. Simulating Sales Data

Получился массив размером 364 × 4: 364 дня и 4 товара. Каждая строка содержит продажи за один день, а каждый столбец — продажи отдельного товара.

## 2. Computing Monthly Averages

Для расчёта условных месяцев берём первые 360 дней: 12 месяцев по 30 дней. Перестраиваем массив в форму «месяц × дни × товары» и находим средние продажи каждого товара за каждый месяц.

```python
monthly_sales = daily_sales[:360].reshape(12, 30, num_products)

monthly_avg_sales = monthly_sales.mean(axis=1)

print("Размер массива средних продаж:", monthly_avg_sales.shape)
print("Строки — месяцы, столбцы — товары:")
print(np.round(monthly_avg_sales, 2))
```

Результат имеет размер 12 × 4. Каждая строка соответствует месяцу, а числа в ней — средние ежедневные продажи четырёх товаров за этот месяц.

## 3. Performance Comparison

Сравним суммирование всех продаж с помощью вложенных циклов Python и функции `np.sum`. Время зависит от компьютера и может меняться между запусками. Для более надёжного сравнения каждый способ выполняется много раз.

```python
def sum_with_loop(data):
    total = 0
    for row in data:
        for item in row:
            total += item
    return total


loop_total = sum_with_loop(daily_sales)
numpy_total = np.sum(daily_sales)

print("Сумма через циклы Python:", loop_total)
print("Сумма через NumPy:", numpy_total)
print("Суммы совпадают:", loop_total == numpy_total)

number_of_runs = 100

loop_time = timeit.timeit(
    lambda: sum_with_loop(daily_sales),
    number=number_of_runs
)

numpy_time = timeit.timeit(
    lambda: np.sum(daily_sales),
    number=number_of_runs
)

print(f"\nВремя циклов Python ({number_of_runs} запусков): {loop_time:.6f} секунд")
print(f"Время np.sum ({number_of_runs} запусков): {numpy_time:.6f} секунд")
print(f"Отношение времени Python-циклов к NumPy: {loop_time / numpy_time:.2f}x")
```

Оба способа вычисляют одинаковую сумму. В этом сравнении `np.sum` обычно работает быстрее, потому что NumPy выполняет вычисления над массивом оптимизированным кодом. Точное время и разница зависят от компьютера.

## 4. Weekly Sales Aggregation

364 дня составляют ровно 52 недели по 7 дней. Перестраиваем данные в форму «неделя × дни × товары», затем складываем продажи за семь дней каждой недели.

```python
weekly_sales_data = daily_sales.reshape(52, 7, num_products)

total_weekly_sales = weekly_sales_data.sum(axis=1)

print("Размер массива недельных итогов:", total_weekly_sales.shape)
print("Продажи четырёх товаров за первую неделю:")
print(total_weekly_sales[0])
```

Результат имеет размер 52 × 4. Каждая строка соответствует неделе, а числа в ней — суммарным продажам каждого товара за эту неделю. В отличие от расчёта среднемесячных продаж, здесь используются все 364 дня.
